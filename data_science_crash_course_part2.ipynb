# NumPy 
NumPy provides ndarrays (n-dimensional arrays)—powerful data structures designed for fast data handling and manipulation.

## Create Array

```python
# define an array
import numpy as np
mylist = [1, 2, 3]
myarray = np.array(mylist)
print(myarray)
print(myarray.shape)
```

## Access Data

Array notation and ranges can be used to efficiently access data in a NumPy array.

```python
# access values
mylist = [[1, 2, 3], [3, 4, 5]]
myarray = np.array(mylist)
print(myarray)
print(myarray.shape)
print("First row: %s" % myarray[0])
print("Last row: %s" % myarray[-1])
print("Specific row and col: %s" % myarray[0, 2])
print("Whole col: %s" % myarray[:, 2])
```

## Arithmetic

NumPy arrays can be used directly in arithmetic.

```python
# arithmetic
myarray1 = np.array([2, 2, 2])
myarray2 = np.array([3, 3, 3])
print("Addition: %s" % (myarray1 + myarray2))
print("Multiplication: %s" % (myarray1 * myarray2))
```

There is a lot more to NumPy arrays but these examples give you a flavor of the efficiencies they provide when working with lots of numerical data.

# Matplotlib Crash Course

Matplotlib can be used for creating plots and charts.

## Line Plot

The example below creates a simple line plot from one-dimensional data.

```python
# basic line plot
import matplotlib.pyplot as plt
myarray = np.array([1, 2, 3])
plt.plot(myarray)
plt.xlabel('some x axis')
plt.ylabel('some y axis')
plt.show()
```

## Scatter Plot

Below is a simple example of creating a scatter plot from two-dimensional data.

```python
# basic scatter plot
x = np.array([1, 2, 3])
y = np.array([2, 4, 6])
plt.scatter(x,y)
plt.xlabel('some x axis')
plt.ylabel('some y axis')
plt.show()
```

# Pandas Crash Course

Pandas provides data structures and functionality to quickly manipulate and analyze data. Instead of manually building DataFrames from scratch, data scientists typically load existing data from files like CSVs.

## Reading a CSV File

Let's read the `Cartwheeldata.csv` file and inspect its contents using `head()`, which shows the first 5 rows of the dataset.

```python
import pandas as pd

# Read the CSV file into a Pandas DataFrame
df = pd.read_csv('Cartwheeldata.csv')

# Display the first 5 rows
df.head()
```

## Data Summary & Statistics

Pandas makes it incredibly easy to compute summary statistics for your data. We can use `.describe()` to get a quick statistical summary of all numeric columns, including count, mean, standard deviation, min, max, and percentiles (like the median at 50%).

```python
# Get summary statistics for the dataset
df.describe()
```

If you only want specific statistics, like the mean or median of a specific column, you can call those methods directly:

```python
# Calculate the mean of the 'Age' column
print("Mean Age:", df['Age'].mean())

# Calculate the median of the 'Score' column
print("Median Score:", df['Score'].median())

# Calculate the mean for all numeric columns
df.mean(numeric_only=True)
```
