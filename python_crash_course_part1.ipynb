## Python Crash Course

When getting started in Python you need to know a few key details about the language syntax to be able to read and understand Python code. This includes:

* Assignment
* Flow Control
* Data Structures
* Functions
* Classes

We will cover each of these topics in turn with small standalone examples that you can type and run.

## Assignment


### Strings

```python
# Strings
data = 'hello world'
print(data[0])
print(len(data))
print(data)
```

### Numbers

```python
# Numbers
value = 123.1
print(value)
value = 10
print(value)
```

### Boolean

```python
# Boolean
a = True
b = False
print(a, b)
```

### Multiple Assignment

```python
# Multiple Assignment
a, b, c = 1, 2, 3
print(a, b, c)
```

### No Value

```python
# No value
a = None
print(a)
```

## Flow Control

There are three main types of flow control that you need to learn: If-Then-Else conditions, For-Loops and While-Loops.

### If-Then-Else Condition Example

```python
value = 99
if value >= 99:
	print('That is fast')
elif value > 200:
	print('That is too fast')
else:
	print('That that is safe')
```

### For-Loop Example

```python
# For-Loop
for i in range(10):
	print(i)
```

### While-Loop Example

```python
# While-Loop
i = 0
while i < 10:
	print(i)
	i += 1
```

## Data Structures

There are three data structures in Python that you will find the most used and useful. They are tuples, lists and dictionaries.

### Tuple Example

Tuples are read-only collections of items.

```python
a = (1, 2, 3)
print(a)
```

### List Example

Lists use the square bracket notation and can be indexed using array notation.

```python
mylist = [1, 2, 3]
print("Zeroth Value: %d" % mylist[0])
mylist.append(4)
print("List Length: %d" % len(mylist))
for value in mylist:
	print(value)
```

### Dictionary Example

Dictionaries are mappings of names to values, like a map. Note the use of the curly bracket notation.

```python
mydict = {'a': 1, 'b': 2, 'c': 3}
print("A value: %d" % mydict['a'])
mydict['a'] = 11
print("A value: %d" % mydict['a'])
print("Keys: %s" % mydict.keys())
print("Values: %s" % mydict.values())
for key in mydict.keys():
	print(mydict[key])
```

The example below defines a new function to calculate the sum of two values and calls the function with two arguments.

## Classes

Classes are the foundation of Object-Oriented Programming in Python. They provide a means of bundling data and functionality together. Creating a new class creates a new type of object, allowing new instances of that type to be made.

### Class Example

In this example, we define a basic `Person` class with an initialization method (`__init__`) and a greeting method.

```python
# Class definition
class Person:
	def __init__(self, name, age):
		self.name = name
		self.age = age

	def greet(self):
		print("Hello, my name is " + self.name)

# Create an object
p1 = Person("Sam", 30)
p1.greet()
```
