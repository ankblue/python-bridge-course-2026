### Exercise 3: Dog Class
Create a class called `Dog` with an `__init__` method that sets a `name` and `breed`, and a `bark` method that prints "Woof!". Instantiate a dog and call the bark method.

```python
class Dog:
    def __init__(self, name, breed):
        self.name = name
        self.breed = breed
        
    def bark(self):
        print("Woof!")

# Instantiate and test
my_dog = Dog("Buddy", "Golden Retriever")
print(f"My dog's name is {my_dog.name} and he is a {my_dog.breed}.")
my_dog.bark()
```

## Part 2: Data Science Libraries Solutions

*First, let's load the necessary libraries and the dataset.*

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv('Cartwheeldata.csv')
```

### Exercise 1: Display Last 3 Rows
Use Pandas to display the last 3 rows of the dataset.

```python
# Display the last 3 rows using .tail()
df.tail(3)
```

### Exercise 2: Maximum Wingspan
Calculate the maximum `Wingspan` from the dataset.

```python
# Calculate the maximum Wingspan
max_wingspan = df['Wingspan'].max()
print("Maximum Wingspan:", max_wingspan)
```

### Exercise 3: Scatter Plot
Create a scatter plot using `Height` for the x-axis and `Wingspan` for the y-axis from the DataFrame.

```python
# Create a scatter plot
plt.scatter(df['Height'], df['Wingspan'], color='blue', alpha=0.7)
plt.title('Height vs Wingspan')
plt.xlabel('Height')
plt.ylabel('Wingspan')
plt.grid(True)
plt.show()
```
